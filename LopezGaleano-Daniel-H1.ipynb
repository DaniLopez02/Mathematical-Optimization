## 1. General Mathematical Formulation

To construct a scalable optimization model for any set of advertisers and search queries, we formulate the Linear Program (LP) using indexed sets, parameters, and decision variables.

### Sets
* $I$: Set of advertisers, indexed by $i \in I$ (Company A, B, C, D).
* $J$: Set of search queries, indexed by $j \in J$ (Q1, Q2, Q3, Q4).

### Parameters
* $c_{ij}$: Average revenue per impression for advertiser $i$ on query $j$, calculated as $\text{PPC}_{ij} \times \text{CTR}_{ij}$ (€).
* $B_i$: Maximum daily budget available for advertiser $i$ (€).
* $R_j$: Estimated maximum number of search requests available per day for query $j$.

### Decision Variables
* $x_{ij} \ge 0$: Number of ad impressions allocated to advertiser $i$ for query $j$.

### Objective Function & Constraints
$$\begin{aligned}
\max_{x_{ij}} \quad & z = \sum_{i \in I} \sum_{j \in J} c_{ij} x_{ij} \quad && \text{(Maximize Google's Total Revenue)} \\
\text{s.t.} \quad & \sum_{j \in J} c_{ij} x_{ij} \le B_i \quad \forall i \in I \quad && \text{(Advertiser Budget Constraints)} \\
& \sum_{i \in I} x_{ij} \le R_j \quad \forall j \in J \quad && \text{(Query Search Request Constraints)} \\
& x_{ij} \ge 0 \quad \forall i \in I, \forall j \in J \quad && \text{(Non-negativity Constraints)}
\end{aligned}$$

## 2. Explicit Problem Formulation (Homework 1 Data)


### Decision Variables
$$x_{A1}, x_{A2}, x_{A3}, x_{A4}, x_{B1}, x_{B2}, x_{B3}, x_{B4}, x_{C1}, x_{C2}, x_{C3}, x_{C4}, x_{D1}, x_{D2}, x_{D3}, x_{D4} \ge 0$$

### Objective Function
$$\begin{aligned}
\max z = & (1.00 x_{A1} + 0.75 x_{A2} + 5.00 x_{A3} + 2.00 x_{A4}) + \\
& (0.50 x_{B1} + 0.50 x_{B2} + 2.00 x_{B3} + 3.00 x_{B4}) + \\
& (0.50 x_{C1} + 3.00 x_{C2} + 1.00 x_{C3} + 0.50 x_{C4}) + \\
& (2.00 x_{D1} + 1.00 x_{D2} + 3.00 x_{D3} + 1.00 x_{D4})
\end{aligned}$$

### Constraints

#### 1. Advertiser Budget Limits ($B_i$)
* **Company A Budget:** $1.00 x_{A1} + 0.75 x_{A2} + 5.00 x_{A3} + 2.00 x_{A4} \le 200$
* **Company B Budget:** $0.50 x_{B1} + 0.50 x_{B2} + 2.00 x_{B3} + 3.00 x_{B4} \le 150$
* **Company C Budget:** $0.50 x_{C1} + 3.00 x_{C2} + 1.00 x_{C3} + 0.50 x_{C4} \le 180$
* **Company D Budget:** $2.00 x_{D1} + 1.00 x_{D2} + 3.00 x_{D3} + 1.00 x_{D4} \le 250$

#### 2. Query Search Demand Limits ($R_j$)
* **Query 1 (Q1 Demand):** $x_{A1} + x_{B1} + x_{C1} + x_{D1} \le 150$
* **Query 2 (Q2 Demand):** $x_{A2} + x_{B2} + x_{C2} + x_{D2} \le 90$
* **Query 3 (Q3 Demand):** $x_{A3} + x_{B3} + x_{C3} + x_{D3} \le 80$
* **Query 4 (Q4 Demand):** $x_{A4} + x_{B4} + x_{C4} + x_{D4} \le 120$

#### 3. Non-negativity Constraints
$$x_{ij} \ge 0 \quad \forall i \in \{A, B, C, D\}, \quad j \in \{1, 2, 3, 4\}$$

## 3. Standard Form with Slack Variables ($s_k$)

Converting all inequality constraints ($\le$) into equalities ($=$) by introducing non-negative slack variables $s_1, s_2, \dots, s_8 \ge 0$:

$$\begin{aligned}
\max z = & \sum_{i \in I} \sum_{j \in J} c_{ij} x_{ij} + \sum_{k=1}^{8} 0 \cdot s_k \\
\text{s.t.} \quad & 1.00 x_{A1} + 0.75 x_{A2} + 5.00 x_{A3} + 2.00 x_{A4} + s_1 = 200 \quad && \text{(Company A Unused Budget)} \\
& 0.50 x_{B1} + 0.50 x_{B2} + 2.00 x_{B3} + 3.00 x_{B4} + s_2 = 150 \quad && \text{(Company B Unused Budget)} \\
& 0.50 x_{C1} + 3.00 x_{C2} + 1.00 x_{C3} + 0.50 x_{C4} + s_3 = 180 \quad && \text{(Company C Unused Budget)} \\
& 2.00 x_{D1} + 1.00 x_{D2} + 3.00 x_{D3} + 1.00 x_{D4} + s_4 = 250 \quad && \text{(Company D Unused Budget)} \\
& x_{A1} + x_{B1} + x_{C1} + x_{D1} + s_5 = 150 \quad && \text{(Unused Searches Q1)} \\
& x_{A2} + x_{B2} + x_{C2} + x_{D2} + s_6 = 90 \quad && \text{(Unused Searches Q2)} \\
& x_{A3} + x_{B3} + x_{C3} + x_{D3} + s_7 = 80 \quad && \text{(Unused Searches Q3)} \\
& x_{A4} + x_{B4} + x_{C4} + x_{D4} + s_8 = 120 \quad && \text{(Unused Searches Q4)} \\
& x_{ij} \ge 0, \quad s_k \ge 0 \quad \forall i, j, k
\end{aligned}$$

```python
import pyomo.environ as pyo

# Model setup
model = pyo.ConcreteModel(name="Google_Ads_Homework_1")

# Decision variables for each company and query (non-negative)
model.x_A1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_A2 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_A3 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_A4 = pyo.Var(domain=pyo.NonNegativeReals)

model.x_B1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_B2 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_B3 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_B4 = pyo.Var(domain=pyo.NonNegativeReals)

model.x_C1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_C2 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_C3 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_C4 = pyo.Var(domain=pyo.NonNegativeReals)

model.x_D1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_D2 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_D3 = pyo.Var(domain=pyo.NonNegativeReals)
model.x_D4 = pyo.Var(domain=pyo.NonNegativeReals)

# Objective: maximize total revenue (PPC * CTR per impression)
model.TotalRevenue = pyo.Objective(
    expr = (
        1.00 * model.x_A1 + 0.75 * model.x_A2 + 5.00 * model.x_A3 + 2.00 * model.x_A4 +
        0.50 * model.x_B1 + 0.50 * model.x_B2 + 2.00 * model.x_B3 + 3.00 * model.x_B4 +
        0.50 * model.x_C1 + 3.00 * model.x_C2 + 1.00 * model.x_C3 + 0.50 * model.x_C4 +
        2.00 * model.x_D1 + 1.00 * model.x_D2 + 3.00 * model.x_D3 + 1.00 * model.x_D4
    ),
    sense = pyo.maximize
)

# Max daily budget per company
model.budget_A = pyo.Constraint(expr = 1.00 * model.x_A1 + 0.75 * model.x_A2 + 5.00 * model.x_A3 + 2.00 * model.x_A4 <= 200)
model.budget_B = pyo.Constraint(expr = 0.50 * model.x_B1 + 0.50 * model.x_B2 + 2.00 * model.x_B3 + 3.00 * model.x_B4 <= 150)
model.budget_C = pyo.Constraint(expr = 0.50 * model.x_C1 + 3.00 * model.x_C2 + 1.00 * model.x_C3 + 0.50 * model.x_C4 <= 180)
model.budget_D = pyo.Constraint(expr = 2.00 * model.x_D1 + 1.00 * model.x_D2 + 3.00 * model.x_D3 + 1.00 * model.x_D4 <= 250)

# Max estimated search volume per query
model.demand_Q1 = pyo.Constraint(expr = model.x_A1 + model.x_B1 + model.x_C1 + model.x_D1 <= 150)
model.demand_Q2 = pyo.Constraint(expr = model.x_A2 + model.x_B2 + model.x_C2 + model.x_D2 <= 90)
model.demand_Q3 = pyo.Constraint(expr = model.x_A3 + model.x_B3 + model.x_C3 + model.x_D3 <= 80)
model.demand_Q4 = pyo.Constraint(expr = model.x_A4 + model.x_B4 + model.x_C4 + model.x_D4 <= 120)

# Import dual values (shadow prices)
model.dual = pyo.Suffix(direction=pyo.Suffix.IMPORT)

# Solve using GLPK
solver = pyo.SolverFactory('glpk')
results = solver.solve(model)

# Print summary
print("Homework 1 - Google Ads Revenue Optimization")
print(f"Status: {results.solver.status}")
print(f"Max Revenue: {pyo.value(model.TotalRevenue):,.2f} €\n")

print("Active Decision Variables:")
variables = [
    ("x_A1 (Company A - Q1)", model.x_A1), ("x_A2 (Company A - Q2)", model.x_A2), 
    ("x_A3 (Company A - Q3)", model.x_A3), ("x_A4 (Company A - Q4)", model.x_A4),
    ("x_B1 (Company B - Q1)", model.x_B1), ("x_B2 (Company B - Q2)", model.x_B2), 
    ("x_B3 (Company B - Q3)", model.x_B3), ("x_B4 (Company B - Q4)", model.x_B4),
    ("x_C1 (Company C - Q1)", model.x_C1), ("x_C2 (Company C - Q2)", model.x_C2), 
    ("x_C3 (Company C - Q3)", model.x_C3), ("x_C4 (Company C - Q4)", model.x_C4),
    ("x_D1 (Company D - Q1)", model.x_D1), ("x_D2 (Company D - Q2)", model.x_D2), 
    ("x_D3 (Company D - Q3)", model.x_D3), ("x_D4 (Company D - Q4)", model.x_D4)
]

# Print non-zero values
for name, var in variables:
    val = pyo.value(var)
    if val > 0:
        print(f"  {name:<25}: {val:6.1f} impressions")

print("\nBudget Shadow Prices:")
print(f"  Company A: {model.dual[model.budget_A]:.2f} €/€")
print(f"  Company B: {model.dual[model.budget_B]:.2f} €/€")
print(f"  Company C: {model.dual[model.budget_C]:.2f} €/€")
print(f"  Company D: {model.dual[model.budget_D]:.2f} €/€")

print("\nQuery Demand Shadow Prices:")
print(f"  Query Q1: {model.dual[model.demand_Q1]:.2f} €/request")
print(f"  Query Q2: {model.dual[model.demand_Q2]:.2f} €/request")
print(f"  Query Q3: {model.dual[model.demand_Q3]:.2f} €/request")
print(f"  Query Q4: {model.dual[model.demand_Q4]:.2f} €/request")
```
